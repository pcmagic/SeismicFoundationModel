```python
import segyio
import numpy as np
import matplotlib.pyplot as plt

# filename = "/home/zhangji/sdb1_data/Geophysics_data/NOPIMS/P00711885/T903_200_EXMOUTH_MC_BHP_LINES_2DTieLine001.SEGY"  # 替换为你的文件名
# filename = "/home/zhangji/sdb1_data/Geophysics_data/NOPIMS/P00711885/T903_200_EXMOUTH_MC_BHP_LINES_XLINE6281.SEGY"  # 替换为你的文件名
filename = "/home/zhangji/sdb1_data/Geophysics_data/NOPIMS/P00590616/FILE_000002.sgy"  # 替换为你的文件名

with segyio.open(filename, 'r', ignore_geometry=True) as segyfile:
    segyfile.mmap()
    print("总地震道数量和时间采样点数")
    print(segyfile.trace)
    print("总地震道数量：")
    print(segyfile.tracecount)
    print("采样时间点数量：")
    print(len(segyfile.samples))
#     print("每炮的地震道数量：")
#     print(segyfile.header[-1][segyio.TraceField.TraceNumber])
#     print("炮的数量：")
#     print(segyfile.tracecount / segyfile.header[-1][segyio.TraceField.TraceNumber])
    data = []
    for i in range(segyfile.tracecount):
        data.append(segyfile.trace[i])
    data = np.array(data).T
    vlim = max(abs(np.max(data)), abs(np.min(data))) / 10
    plt.imshow(data, cmap=plt.cm.seismic, vmax=vlim, vmin=-vlim, aspect="auto")
    plt.colorbar()
    plt.show()
```
